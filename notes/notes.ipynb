

## Geometry notes

Since we took cylindrical represnetation to parametrize the cassegrain geometry we have the following equation that describe the perfect parabolic surface with focus f:

$$S_0(r, \phi) = \left(r\cos(\phi), r\sin(\phi), \frac{r^2}{4f} \right)$$


Then we could compute the tangent vectors of that surface as:

$$S_{0,r} = \frac{\partial S_0}{\partial r} = \left(\cos(\phi), \sin(\phi), \frac{r}{2f} \right)$$

$$S_{0,\phi} = \frac{\partial S_0}{\partial \phi} = \left(-r\sin(\phi), r\cos(\phi),0 \right)$$

To normalize them:
$$||S_{0,r}|| = \sqrt{1+\frac{r^2}{4f^2}}$$

$$||S_{0,\phi}|| = r$$

And with that we can build the tangent coordinate system as:

$$\hat{e}_{r} = \frac{S_{0,r}}{||S_{0,r}||} = \left( \cos(\phi), \sin(\phi), \frac{r}{2f} \right) \frac{1}{\sqrt{1+\frac{r^2}{4f^2}}}$$

$$\hat{e}_{\phi} = \frac{S_{0,\phi}}{||S_{0,\phi}||} = \left(-\sin(\phi), \cos(\phi),0 \right) $$


Now we can compute the normal vectors as:

$$ \hat{n} = \hat{e}_r \times \hat{e}_{\phi} = \left( \frac{-r^2}{2f} \cos(\phi), \frac{-r^2}{2f}\sin(\phi), r \right) \cdot \frac{1}{\sqrt{1+\frac{r^2}{4f^2}}}$$


For the differential surface:
$$ dS = (S_{0,r}\times S_{0,\phi}) dr d\phi =  r \sqrt{1+\frac{r^2}{4f^2}}$$

Now to add the local deformations we have to define a local coordinate system around $p_0$

$$x' = \left(p - p_0\right)\cdot \hat{e}_{r} $$
$$y' = \left(p - p_0\right)\cdot \hat{e}_{\phi} $$


Then the modified surface will be:
$$ S = S_{0}(r,\phi)+f(x'(r, \phi), y'(r,\phi))\cdot \hat{n}_0(r,\phi)$$

Where $f(x',y') = a+bx'+cy'+d(x'^2+y'^2)+e(x'^2-y'^2)$.

And once again we can compute the tangent vectors, the normal vector and the differential surface following the previous recipe.. for that we will need to compute some expression before hand:

$$ \frac{\partial \hat{e}_r}{\partial r} = \left( \frac{\cos({\phi})r}{2f^2\left(1+r^2/(4f^2)\right)^{3/2}}, \frac{\sin({\phi})r}{2f^2\left(1+r^2/(4f^2)\right)^{3/2}}, \frac{1}{2f^2 \sqrt{1+r^2/(4f^2)}}+\frac{r^2}{4f^3 \left(1+r^2/(4f^2)\right)^{3/2}} \right)$$


$$ \frac{\partial \hat{e}_r}{\partial \phi} = (-\sin(\phi), \cos(\phi), 0) \frac{1}{\sqrt{1+\frac{r^2}{4f^2}}}$$


$$ \frac{\partial \hat{e}_\phi}{\partial r} = (0,0,0)$$
$$ \frac{\partial \hat{e}_\phi}{\partial \phi} = (-\cos(\phi), -\sin(\phi), 0)$$

For the next part we note that:
$$S_{0,r}\cdot \hat{e}_r = S_{0,r} \cdot \frac{S_{0,r}}{||S_{0,r}||} = ||S_{0,r}||$$
$$S_{0,r} \cdot \hat{e}_\phi = S_{0,r} \cdot \frac{S_{0,\phi}}{||S_{0,\phi}||} = \left(\cos(\phi), \sin(\phi), \frac{r}{2f} \right) \cdot \left(-r\sin(\phi), r\cos(\phi), 0\right) \frac{1}{||S_{0,\phi}||} = 0$$

$$S_{0,\phi} \cdot \hat{e}_r = 0 $$

$$S_{0,\phi}\cdot \hat{e}_\phi = ||S_{0,\phi} || $$


We also will need:

$$ \frac{\partial x'}{\partial r} = \frac{\partial p}{\partial r} \hat{e}_{r}+ \left(p-p_0 \right)\frac{\partial \hat{e}_r}{\partial r} = \frac{\partial S(r,\phi)}{\partial r} \hat{e}_{r}+ \left(p-p_0 \right)\frac{\partial \hat{e}_r}{\partial r} = S_{0,r} \cdot \hat{e}_r+\left(p-p_0 \right) \cdot \frac{\partial \hat{e}_r}{\partial r}$$


$$ \frac{\partial x'}{\partial \phi} = S_{0,\phi} \cdot \hat{e}_r+ \left(p-p_0\right) \cdot \frac{\partial \hat{e}_r}{\partial \phi} = \left(p-p_0\right) \cdot \frac{\partial \hat{e}_r}{\partial \phi} $$


$$ \frac{\partial y'}{\partial r} = S_{0, r} \cdot \hat{e}_\phi+\left(p-p_0 \right)\cdot \frac{\partial \hat{e}_\phi}{\partial r} = 0$$


$$ \frac{\partial y'}{\partial \phi} = S_{0, \phi} \cdot \hat{e}_\phi+ \left(p-p0 \right) \cdot \frac{\partial \hat{e}_\phi}{\partial \phi}$$



Ok, now we have that the tangent vectors are:

$$S_{r} = \frac{\partial S}{\partial r} = S_{0,r}+\left[\frac{\partial f}{\partial x'} \frac{\partial x'}{\partial r}+ \frac{\partial f}{\partial y'} \frac{\partial y'}{\partial r} \right] \hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial r}$$

$$S_{\phi} = \frac{\partial S}{\partial \phi} = S_{0,\phi}+\left[\frac{\partial f}{\partial x'} \frac{\partial x'}{\partial \phi}+ \frac{\partial f}{\partial y'} \frac{\partial y'}{\partial \phi} \right] \hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial \phi}$$







To make it more awfull Ill expand the expressions:

$$ S_{r} = S_{0,r} + \left[  \frac{\partial f}{\partial x'} \left( S_{0,r} \cdot \hat{e}_r + (\vec{p}-\vec{p}_0)\cdot \frac{\partial \hat{e}_r}{\partial r}  \right) + \frac{\partial f}{\partial y'} \left( S_{0,r}\cdot \hat{e}_\phi \right) \right] \hat{n}_0  + f(x',y') \frac{\partial \hat{n}_0}{\partial r}$$

$$ S_r \stackrel{?}{=} S_{0,r} + \frac{\partial f}{\partial x'} \left[||S_{0,r}||+ (\vec{p}-\vec{p}_0)\cdot \frac{\partial \hat{e}_r}{\partial r} \right] \hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial r}$$




$$S_\phi = S_{0,\phi} + \left[ \frac{\partial f}{\partial x'} \left(S_{0,\phi}\cdot \hat{e}_r + (\vec{p}-\vec{p}_0) \cdot \frac{\partial \hat{e}_r}{\partial \phi} \right) + \frac{\partial f}{\partial y'} \left(S_{0,\phi}\cdot \hat{e}_\phi + (\vec{p}-\vec{p}_0)\cdot \frac{\partial \hat{e}_\phi}{\partial \phi}\right) \right] \hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial \phi}$$

$$S_\phi \stackrel{?}{=} S_{0, \phi} + \left( (\vec{p}-\vec{p}_0)\cdot\left[ \frac{\partial f}{\partial x'}\frac{\partial \hat{e}_r}{\partial \phi} + \frac{\partial f}{\partial y'} \frac{\partial \hat{e}_\phi}{\partial \phi}\right] + \frac{\partial f}{\partial y'}||S_{0,\phi} || \right) \hat{n}_0+ f(x',y')\frac{\partial \hat{n}_0}{\partial \phi}$$


With that the normal vector is (not normalized..):
$$\vec{n} = S_r \times S_\phi $$

After that we build the differential surface as:
$$dS =  (S_{r} \times S_{\phi}) dr d\phi$$
