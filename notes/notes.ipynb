

## Geometry notes

Since we took cylindrical represnetation to parametrize the cassegrain geometry we have the following equation that describe the perfect parabolic surface with focus f:

$$S_0(r, \phi) = \left(r\cos(\phi), r\sin(\phi), \frac{r^2}{4f} \right)$$


Then we could compute the tangent vectors of that surface as:

$$S_{0,r} = \frac{\partial S_0}{\partial r} = \left(\cos(\phi), \sin(\phi), \frac{r}{2f} \right)$$

$$S_{0,\phi} = \frac{\partial S_0}{\partial \phi} = \left(-r\sin(\phi), r\cos(\phi),0 \right)$$

To normalize them:
$$||S_{0,r}|| = \sqrt{1+\frac{r^2}{4f^2}} = \rho$$

$$||S_{0,\phi}|| = r$$

And with that we can build the tangent coordinate system as:

$$\hat{e}_{r} = \frac{S_{0,r}}{||S_{0,r}||} = \left( \cos(\phi), \sin(\phi), \frac{r}{2f} \right) \frac{1}{\sqrt{1+\frac{r^2}{4f^2}}}$$

$$\hat{e}_{\phi} = \frac{S_{0,\phi}}{||S_{0,\phi}||} = \left(-\sin(\phi), \cos(\phi),0 \right) $$


Now we can compute the normal vectors as:

$$ \hat{n}_0 = \hat{e}_r \times \hat{e}_{\phi} = \left( -\frac{r}{2f} \cos(\phi), -\frac{r}{2f}\sin(\phi), 1 \right) \cdot \frac{1}{\sqrt{1+\frac{r^2}{4f^2}}}$$


For the differential surface:
$$ dS = (S_{0,r}\times S_{0,\phi}) dr d\phi =  r \sqrt{1+\frac{r^2}{4f^2}} dr d\phi$$

Now to add the local deformations we have to define a local coordinate system around $p_0$

$$x' = \left(p - p_0\right)\cdot \hat{e}_{r} $$
$$y' = \left(p - p_0\right)\cdot \hat{e}_{\phi} $$


Then the modified surface will be:
$$ S = S_{0}(r,\phi)+f(x'(r, \phi), y'(r,\phi))\cdot \hat{n}_0(r,\phi)$$

Where $f(x',y') = a+bx'+cy'+d(x'^2+y'^2)+e(x'^2-y'^2)$.

And once again we can compute the tangent vectors, the normal vector and the differential surface following the previous recipe.. for that we will need to compute some expression before hand:

$$ \frac{\partial}{\partial r} \left(\frac{1}{\rho} \right) = -\frac{1}{\rho^2} \frac{\partial \rho}{\partial r} = -\frac{1}{\rho^2} \frac{1}{2\sqrt{1+\frac{r^2}{4f^2}}} \frac{r}{2f^2} = -\frac{1}{\rho^3}\frac{r}{4f^2} $$

Then:
$$\frac{\partial \hat{e}_r}{\partial r} = \left(-\frac{r \cos(\phi)}{4f^2 \rho^3} , -\frac{r \sin(\phi)}{4f^2 \rho^3}, \frac{1}{2f\rho} -\frac{r^2}{8f^3 \rho^3}\right)$$

$$ \frac{\partial \hat{e}_r}{\partial \phi} = \left(-\frac{\sin(\phi)}{\rho}, \frac{\cos(\phi)}{\rho}, 0 \right) $$

$$ \frac{\partial \hat{e}_\phi}{\partial r} = 0$$

$$ \frac{\partial \hat{e}_\phi}{\partial \phi} = \left( -\cos(\phi), -\sin(\phi), 0 \right) $$



We also note that:

$$S_{0,r}\cdot \hat{e}_r = S_{0,r} \cdot \frac{S_{0,r}}{||S_{0,r}||} = ||S_{0,r}|| = \rho $$

$$S_{0,r}\cdot \hat{e}_\phi = \left( \cos(\phi), \sin(\phi), \frac{r}{2f} \right) \cdot \left(-\sin(\phi), \cos(\phi), 0 \right) = 0$$

$$ S_{0,\phi} \cdot \hat{e}_r = \left(-r\sin(\phi), r\cos(\phi), 0 \right) \cdot \left( \cos(\phi), \sin(\phi), \frac{r}{2f} \right) \frac{1}{\rho} = 0 $$

$$S_{0,\phi} \cdot \hat{e}_\phi = ||S_{0,\phi}|| = r$$



And we will need:
$$ \frac{\partial x'}{\partial r} =  S_{0,r}\cdot \hat{e}_r + (p-p_0)\cdot \frac{\partial \hat{e}_r}{\partial r} = \rho+(p-p_0)\cdot \frac{\partial \hat{e}_r}{\partial r} $$

$$ \frac{\partial x'}{\partial \phi} = S_{0,\phi} \cdot \hat{e}_r + (p-p_0) \cdot \frac{\hat{e}_r}{\partial \phi} = (p-p_0) \cdot \frac{\hat{e}_r}{\partial \phi}$$

$$\frac{\partial y'}{\partial r} = S_{0,r} \cdot \hat{e}_\phi + (p-p_0)\cdot \frac{\partial \hat{e}_\phi}{\partial r} = 0 $$

$$\frac{\partial y'}{\partial \phi} = S_{0,\phi} \cdot \hat{e}_\phi + (p-p_0) \cdot \frac{\partial \hat{e}_{\phi}}{\partial \phi}= r+(p-p_0) \cdot \frac{\partial \hat{e}_{\phi}}{\partial \phi}$$

And the last expression that we need are:
$$ \frac{\partial \hat{n}_0}{\partial r} = \left( \frac{\cos(\phi)}{2f\rho} \left[ \frac{r^2}{4f^2\rho^2} -1 \right], \frac{\sin(\phi)}{2f\rho} \left[ \frac{r^2}{4f^2\rho^2} -1 \right], -\frac{r}{4f^2\rho^3}  \right) $$

to further reduce the expression we take a look only in the x component:
$$ (\partial_r \hat{n}_0)_x = \frac{\cos(\phi)}{2f\rho^3} \left(\frac{r^2}{4f^2} -\rho^2  \right) =\frac{\cos(\phi)}{2f\rho^3} \left(\frac{r^2}{4f^2} - \left( 1+\frac{r^2}{4f^2}\right) \right) =-\frac{\cos(\phi)}{2f\rho^3} $$

So:
$$\frac{\partial \hat{n}_0}{\partial r} = \left(-\frac{\cos(\phi)}{2f\rho^3}, -\frac{\sin(\phi)}{2f\rho^3}, -\frac{r}{4f^2\rho^3} \right) $$


$$\frac{\partial \hat{n}_0}{ \partial \phi} = \left( \frac{r\sin(\phi)}{2f\rho}, -\frac{r\cos(\phi)}{2f\rho}, 0 \right) $$

Ok, after all this gibberish we finally can compute:

$$S_{r} = \frac{\partial S}{\partial r} =  S_{0,r} + \left[\frac{\partial f}{\partial x'} \frac{\partial x'}{\partial r}+\frac{\partial f}{\partial y'} \frac{\partial y'}{\partial r} \right]\hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial r} $$ 


$$S_{\phi} = \frac{\partial S}{\partial \phi} =  S_{0,\phi} + \left[\frac{\partial f}{\partial x'} \frac{\partial x'}{\partial \phi}+\frac{\partial f}{\partial y'} \frac{\partial y'}{\partial \phi} \right]\hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial \phi} $$

Reeplacing the previous expressions:
$$S_{r} = S_{0,r}+\frac{\partial f}{\partial x'}\left(\rho+(\vec{p}-\vec{p}_0)\cdot \frac{\partial \hat{e}_r}{\partial r} \right)\hat{n}_0+f(x',y')\frac{\partial \hat{n}_0}{\partial r} $$

$$S_{\phi} = S_{0,\phi} +  \left[\frac{\partial f}{\partial x'}(\vec{p}-\vec{p}_0) \cdot \frac{\partial \hat{e}_r}{\partial \phi}+ \frac{\partial f}{\partial y'} \left( r+(\vec{p}-\vec{p}_0) \cdot \frac{\partial \hat{e}_{\phi}}{\partial \phi} \right) \right]\hat{n}_0 + f(x',y') \frac{\partial \hat{n}_0}{\partial \phi}$$

$$= S_{0,\phi} + \left[ (\vec{p}-\vec{p}_0)\cdot \left( \frac{\partial f}{\partial x'}\frac{\partial \hat{e}_r}{\partial \phi} + \frac{\partial f}{\partial y'} \frac{\partial \hat{e}_\phi}{\partial \phi}\right) +r\frac{\partial f}{\partial y'}\right]\hat{n}_0 + f(x',y')\frac{\partial \hat{n}_0}{\partial \phi} $$

With that the normal vector is (not normalized..):
$$\vec{n} = S_r \times S_\phi $$

After that we build the differential surface as:
$$dS =  (S_{r} \times S_{\phi}) dr d\phi$$
